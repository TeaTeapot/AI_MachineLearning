### Принятые методические решения

Целевая переменная Potability бинарная (0 — непитьевая, 1 — питьевая), поэтому, основной моделью выбрана логистическая регрессия в трёх вариантах: без регуляризации, с L1-штрафом (аналог Lasso) и с L2-штрафом (аналог Ridge). Чтобы выполнить пункт задания буквально, линейная регрессия, Lasso и Ridge дополнительно обучены на целевой переменной, закодированной как 0/1 (т. н. линейная вероятностная модель).

Метрики RMSE, R² и MAPE рассчитаны по предсказанным вероятностям класса 1 (для линейных моделей — по их выходу). Это накладывает ограничения:

* $\text{RMSE}^2$ в этом случае совпадает с оценкой Брайера (Brier score);
* MAPE не определён для объектов класса 0 (деление на ноль), поэтому он считается только по объектам класса 1;
* чтобы оценка была содержательной, дополнительно приведены Accuracy, F1 и ROC-AUC, а все модели сравниваются с baseline — константным прогнозом, равным доле класса 1 в обучающей выборке.

Защита от утечки данных. Стандартизация и PCA обучаются только на обучающей части данных: они помещены внутрь Pipeline и поэтому внутри кросс-валидации видят лишь обучающие блоки. Это важно: подгонка StandardScaler/PCA на всей выборке до разбиения завышает оценку качества.

### Краткая теория

* Линейная регрессия (МНК): $\hat y = w_0 + \sum_j w_j x_j$, веса минимизируют $\sum_i (y_i-\hat y_i)^2$.
* Ridge (гребневая): к функции потерь добавляется $\alpha \sum_j w_j^2$ (L2-штраф), веса сжимаются, но не обнуляются.
* Lasso: добавляется $\alpha \sum_j |w_j|$ (L1-штраф); часть весов обнуляется — встроенный отбор признаков.
* Логистическая регрессия: $P(y=1\mid x)=\sigma(w_0+\sum_j w_j x_j)$; штрафы L1/L2 задаются коэффициентом $C = 1/\alpha$ (чем меньше $C$, тем сильнее регуляризация).
* VIF: $\text{VIF}_j = \dfrac{1}{1-R_j^2}$, где $R_j^2$ — коэффициент детерминации регрессии признака $j$ на остальные признаки. Ориентиры: < 5 — норма, 5–10 — умеренная, > 10 — сильная мультиколлинеарность.
* Метрики: $\text{RMSE}=\sqrt{\frac1n\sum(y_i-\hat y_i)^2}$, $\;R^2=1-\dfrac{\sum(y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}$, $\;\text{MAPE}=\frac1n\sum\left|\dfrac{y_i-\hat y_i}{y_i}\right|$.
* PCA: ортогональное преобразование стандартизованных признаков в некоррелированные главные компоненты, упорядоченные по доле объяснённой дисперсии (собственным значениям ковариационной матрицы).

## 3. Подготовка данных

1. **Пропуски.** В соответствии с заданием строки с пропусками удаляются. Цена этого решения — потеря **38,6 %** наблюдений (3276 → 2011); доля класса 1 при этом почти не изменилась (0,390 → 0,403). Альтернатива (заполнение медианой) проверена в п. 4.8.
2. **Кодирование.** Категориальных признаков нет — не требуется.
3. **Разбиение.** Выборка делится на обучающую (80 %) и тестовую (20 %) функцией `train_test_split` со стратификацией по классу (`random_state=42`): 1608 и 403 наблюдения, доля класса 1 в обеих — 0,40.
4. **Стандартизация.** Выполняется `StandardScaler` **внутри `Pipeline`** — среднее и дисперсия оцениваются только по обучающим данным.
5. **Кросс-валидация.** Используется стратифицированная 5-блочная схема. Гиперпараметры ($C$ или $\alpha$) подбираются по логарифмической сетке из 15 значений во внутреннем цикле (`GridSearchCV`), а качество оценивается во внешнем цикле (**вложенная кросс-валидация**), поэтому оценка не завышается подбором параметров.

```python
n_before = len(df)
df_clean = df.dropna().reset_index(drop=True)
print(f"Строк до удаления пропусков:    {n_before}")
print(f"Строк после удаления пропусков: {len(df_clean)}  (потеряно {(1 - len(df_clean) / n_before) * 100:.1f}%)")
print("Доля класса 1 до / после: "
      f"{df['Potability'].mean():.3f} / {df_clean['Potability'].mean():.3f}")

X = df_clean.drop(columns="Potability")
y = df_clean["Potability"]

# Разброс масштабов признаков — обоснование стандартизации
print("\nСтандартные отклонения признаков (до стандартизации):")
print(X.std().round(2))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)

print(f"Обучающая выборка: {X_train.shape[0]} строк, доля класса 1 = {y_train.mean():.3f}")
print(f"Тестовая выборка:  {X_test.shape[0]} строк, доля класса 1 = {y_test.mean():.3f}")

# Схемы кросс-валидации: внешняя — для оценки качества, внутренняя — для подбора гиперпараметров
outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
inner_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE + 1)
```

### 4.1. Корреляции между признаками и мультиколлинеарность

```python
corr_feat = X_train.corr()
mask = np.triu(np.ones_like(corr_feat, dtype=bool))

fig, ax = plt.subplots(figsize=(8, 6.5))
sns.heatmap(corr_feat, mask=mask, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True, ax=ax)
ax.set_title("Корреляции между признаками (обучающая выборка)")
plt.tight_layout()
plt.show()

off_diag = corr_feat.where(~mask).stack()
print(f"Максимальная попарная |корреляция| признаков: {off_diag.abs().max():.3f} "
      f"({' — '.join(off_diag.abs().idxmax())})")
```

```python
def compute_vif(features: pd.DataFrame) -> pd.Series:
    """VIF_j = 1 / (1 - R_j^2), где R_j^2 — коэффициент детерминации регрессии
    признака j на все остальные признаки (эквивалентно statsmodels.variance_inflation_factor)."""
    result = {}
    for col in features.columns:
        others = features.drop(columns=col)
        r2 = LinearRegression().fit(others, features[col]).score(others, features[col])
        result[col] = 1.0 / (1.0 - r2)
    return pd.Series(result, name="VIF")

vif = compute_vif(X_train)
print(vif.round(3).to_frame())

fig, ax = plt.subplots(figsize=(7, 3.8))
sns.barplot(x=vif.values, y=vif.index, ax=ax)
ax.axvline(5, color="orange", ls="--", label="VIF = 5")
ax.axvline(10, color="red", ls="--", label="VIF = 10")
ax.set_xlim(0, 11)
ax.set_xlabel("VIF")
ax.set_title("Коэффициенты инфляции дисперсии (VIF)")
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

**Вывод.** Максимальная попарная корреляция признаков — всего **0,178** (`Sulfate` — `Solids`), значения VIF лежат в диапазоне **1,002–1,050** — на порядок ниже порога 5. **Мультиколлинеарности в данных нет.**

Важное следствие: обычно PCA применяют, чтобы устранить мультиколлинеарность и сократить число признаков, но здесь устранять нечего. Заранее следует ожидать, что PCA не принесёт выигрыша (проверяется в п. 4.5–4.7).

### 4.2. Метрики и протокол оценки

Ниже определены функции расчёта метрик и оценки моделей. Для каждой модели считаются:

* **CV-оценка** — среднее ± стандартное отклонение по 5 внешним блокам (вложенная кросс-валидация на обучающей выборке);
* **Test-оценка** — финальная модель, обученная на всей обучающей выборке, проверяется на отложенной тестовой (403 наблюдения).

Поскольку тестовая выборка небольшая, **CV-оценке следует доверять больше**: на 403 наблюдениях ROC-AUC имеет заметный случайный разброс.

**Baseline.** Константный прогноз $\hat p = 0{,}403$ даёт RMSE ≈ 0,491, $R^2\approx 0$, MAPE (по классу 1) $=1-0{,}403=0{,}597$, ROC-AUC = 0,5. Модель имеет смысл, только если заметно превосходит эти значения.

```python
def mape_pos(y_true, y_pred):
    """MAPE по объектам класса 1.
    Для класса 0 знаменатель |y_true| равен нулю, поэтому MAPE там не определён."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    m = y_true == 1
    return np.mean(np.abs((y_true[m] - y_pred[m]) / y_true[m]))


def compute_scores(y_true, score):
    """Метрики для непрерывного прогноза `score` (вероятность класса 1 или выход линейной модели)."""
    y_true = np.asarray(y_true)
    label = (score >= 0.5).astype(int)
    return {
        "RMSE": np.sqrt(mean_squared_error(y_true, score)),
        "R2": r2_score(y_true, score),
        "MAPE": mape_pos(y_true, score),
        "Accuracy": accuracy_score(y_true, label),
        "F1": f1_score(y_true, label, zero_division=0),
        "ROC-AUC": roc_auc_score(y_true, score),
    }


def predict_score(model, features):
    """Вероятность класса 1 для классификаторов, прямой прогноз — для регрессоров."""
    if hasattr(model, "predict_proba"):
        return model.predict_proba(features)[:, 1]
    return model.predict(features)


def make_logreg(kind):
    """Логистическая регрессия: без штрафа / L1 (Lasso) / L2 (Ridge).
    Учтено изменение API в scikit-learn 1.8 (параметр penalty объявлен устаревшим)."""
    new_api = tuple(int(p) for p in sklearn.__version__.split(".")[:2]) >= (1, 8)
    if new_api:
        return {"none": LogisticRegression(C=np.inf, max_iter=5000),
                "l1": LogisticRegression(l1_ratio=1, solver="liblinear"),
                "l2": LogisticRegression(l1_ratio=0)}[kind]
    return {"none": LogisticRegression(penalty=None, max_iter=5000),
            "l1": LogisticRegression(penalty="l1", solver="liblinear"),
            "l2": LogisticRegression(penalty="l2")}[kind]


def run_model(pipe, grid=None, scoring=None, data=None):
    """Оценка модели двумя способами:
    1) вложенная кросс-валидация на обучающей выборке (5 блоков; гиперпараметры подбираются
       во внутреннем цикле, поэтому оценка не завышена);
    2) финальное обучение на всей обучающей выборке и проверка на тестовой.
    Возвращает словарь метрик и обученную модель."""
    Xtr, ytr, Xte, yte = data if data is not None else (X_train, y_train, X_test, y_test)
    est = GridSearchCV(pipe, grid, cv=inner_cv, scoring=scoring) if grid else pipe

    cv_rows = []
    for tr, va in outer_cv.split(Xtr, ytr):
        m = clone(est).fit(Xtr.iloc[tr], ytr.iloc[tr])
        cv_rows.append(compute_scores(ytr.iloc[va], predict_score(m, Xtr.iloc[va])))
    cv_mean = pd.DataFrame(cv_rows).mean()
    cv_std = pd.DataFrame(cv_rows).std()

    final = clone(est).fit(Xtr, ytr)
    test = compute_scores(yte, predict_score(final, Xte))
    best = final.best_params_ if grid else {}
    return {"cv": cv_mean, "cv_std": cv_std, "test": pd.Series(test),
            "best": best, "model": final.best_estimator_ if grid else final}


LOGIT_GRID = {"m__C": np.logspace(-3, 2, 15)}
LASSO_GRID = {"m__alpha": np.logspace(-4, 0, 15)}
RIDGE_GRID = {"m__alpha": np.logspace(-2, 4, 15)}
```

```python
results = {}   # (пространство признаков, модель) -> результаты

base_pipe = Pipeline([("m", DummyClassifier(strategy="prior"))])
results[("Константа", "Baseline")] = run_model(base_pipe)
print("Baseline (всегда предсказывает долю класса 1 из обучающей выборки):")
print(results[("Константа", "Baseline")]["test"].round(4).to_frame("test"))
```

### 4.3. Логистическая регрессия на исходных признаках

```python
logit_specs = {
    "Логистическая (без штрафа)": (make_logreg("none"), None),
    "Логистическая + L1 (Lasso)": (make_logreg("l1"), LOGIT_GRID),
    "Логистическая + L2 (Ridge)": (make_logreg("l2"), LOGIT_GRID),
}

for name, (clf, grid) in logit_specs.items():
    pipe = Pipeline([("scaler", StandardScaler()), ("m", clf)])
    results[("Исходные", name)] = run_model(pipe, grid, scoring="neg_log_loss")
    print(f"{name:30s} лучшие гиперпараметры: {results[('Исходные', name)]['best']}")
```

```python
def summary_table(space, names):
    """Сводная таблица: метрики CV (среднее ± std) и на тестовой выборке."""
    rows = []
    for n in names:
        r = results[(space, n)]
        row = {"Модель": n}
        for k in ["RMSE", "R2", "MAPE", "ROC-AUC"]:
            row[f"CV {k}"] = f"{r['cv'][k]:.3f} ± {r['cv_std'][k]:.3f}"
        for k in ["RMSE", "R2", "MAPE", "Accuracy", "F1", "ROC-AUC"]:
            row[f"Test {k}"] = round(r["test"][k], 4)
        rows.append(row)
    return pd.DataFrame(rows).set_index("Модель")

logit_names = list(logit_specs)
summary_table("Константа", ["Baseline"]).T.join(summary_table("Исходные", logit_names).T)
```

```python
coef_tbl = pd.DataFrame({
    n: results[("Исходные", n)]["model"].named_steps["m"].coef_[0] for n in logit_names
}, index=X.columns)
print("Коэффициенты логистических моделей (признаки стандартизованы):")
display(coef_tbl.round(4))
print("Число коэффициентов, обнулённых L1-штрафом:",
      int((coef_tbl["Логистическая + L1 (Lasso)"] == 0).sum()), "из", len(coef_tbl))
```

```python
def plot_roc(space, names, title):
    fig, ax = plt.subplots(figsize=(5.5, 5))
    for n in names:
        s = predict_score(results[(space, n)]["model"], X_test)
        fpr, tpr, _ = roc_curve(y_test, s)
        ax.plot(fpr, tpr, label=f"{n} (AUC = {roc_auc_score(y_test, s):.3f})")
    ax.plot([0, 1], [0, 1], "k--", label="Случайное угадывание")
    ax.set_xlabel("Доля ложноположительных (FPR)")
    ax.set_ylabel("Доля истинноположительных (TPR)")
    ax.set_title(title)
    ax.legend(fontsize=8, loc="lower right")
    plt.tight_layout()
    plt.show()

plot_roc("Исходные", logit_names, "ROC-кривые, логистические модели (исходные признаки)")
```

**Анализ результатов.**

* Все три модели практически **не отличаются от baseline**: CV-RMSE 0,490–0,491 (у baseline 0,490), CV-$R^2$ от −0,002 до 0,000, CV ROC-AUC 0,510–0,514 (у случайного угадывания 0,5). Точность на тесте (0,5955) **совпадает с точностью константного прогноза** — модели почти всегда предсказывают класс 0, а F1 близок к нулю.
* Оптимальное $C$ для L2-модели оказалось на **нижней границе сетки** ($C=0{,}001$): кросс-валидация предпочитает максимально «сжатую» модель, то есть признаки не несут линейного сигнала.
* L1-штраф обнулил **6 из 9** коэффициентов, оставив `Solids`, `Conductivity` и (с крошечным весом) `Sulfate`. Но веса малы (≤ 0,05 при стандартизованных признаках), а ROC-AUC колеблется между блоками на ±0,02, поэтому считать эти признаки «важными» нельзя — это след шума, а не закономерность.
* Чуть меньший MAPE у L1-модели (0,581 против 0,597) не является улучшением: MAPE считается только по классу 1 и «вознаграждает» завышение вероятностей; различительная способность (ROC-AUC) при этом не выросла.

### 4.4. Линейная регрессия, Lasso и Ridge (целевая переменная 0/1)

```python
linear_specs = {
    "Линейная (OLS)": (LinearRegression(), None, None),
    "Lasso": (Lasso(max_iter=20000), LASSO_GRID, "neg_root_mean_squared_error"),
    "Ridge": (Ridge(), RIDGE_GRID, "neg_root_mean_squared_error"),
}

for name, (reg, grid, scoring) in linear_specs.items():
    pipe = Pipeline([("scaler", StandardScaler()), ("m", reg)])
    results[("Исходные", name)] = run_model(pipe, grid, scoring)
    print(f"{name:16s} лучшие гиперпараметры: {results[('Исходные', name)]['best']}")

linear_names = list(linear_specs)
summary_table("Исходные", linear_names).T
```

**Анализ результатов.** Картина та же: RMSE 0,490–0,491, $R^2$ в пределах ±0,006, ROC-AUC ≈ 0,51. Показательно поведение регуляризации: оптимальный $\alpha$ для Lasso ($\approx 0{,}037$) обнуляет **все** коэффициенты, и модель вырождается в константу ($R^2=0$, ROC-AUC = 0,5 ровно), а для Ridge $\alpha$ выбран на верхней границе сетки ($10^4$). Обычная МНК-модель, наоборот, чуть хуже константы на тесте ($R^2=-0{,}0057$) — это небольшое переобучение на шуме. Кроме того, линейная модель может выдавать «вероятности» вне отрезка [0, 1], поэтому для бинарной цели логистическая регрессия методологически предпочтительнее.

### 4.5. Снижение размерности методом главных компонент

PCA применяется к **стандартизованным** признакам и обучается только на обучающей выборке.

```python
# PCA обучаем только на обучающей выборке; перед PCA — стандартизация
scaler_full = StandardScaler().fit(X_train)
Z_train = scaler_full.transform(X_train)

pca_full = PCA(random_state=RANDOM_STATE).fit(Z_train)
eigenvalues = pca_full.explained_variance_
ratio = pca_full.explained_variance_ratio_
cum_ratio = np.cumsum(ratio)

pca_table = pd.DataFrame({
    "Собственное значение": eigenvalues,
    "Доля дисперсии": ratio,
    "Накопленная доля": cum_ratio,
}, index=[f"PC{i}" for i in range(1, len(ratio) + 1)])
print(pca_table.round(4))

n_kaiser = int((eigenvalues > 1).sum())
n_85 = int(np.argmax(cum_ratio >= 0.85) + 1)
print(f"\nКомпонент по критерию Кайзера (λ > 1): {n_kaiser}")
print(f"Компонент для накопленной доли ≥ 85%:  {n_85}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
idx = np.arange(1, len(eigenvalues) + 1)

axes[0].plot(idx, eigenvalues, "o-")
axes[0].axhline(1, color="red", ls="--", label="λ = 1 (критерий Кайзера)")
axes[0].set_title("График каменистой осыпи (scree plot)")
axes[0].set_xlabel("Номер главной компоненты")
axes[0].set_ylabel("Собственное значение")
axes[0].set_xticks(idx)
axes[0].legend()

axes[1].bar(idx, ratio, alpha=0.6, label="Доля дисперсии компоненты")
axes[1].plot(idx, cum_ratio, "ro-", label="Накопленная доля")
axes[1].axhline(0.80, color="gray", ls=":", label="80%")
axes[1].axhline(0.90, color="gray", ls="--", label="90%")
axes[1].set_title("Объяснённая дисперсия")
axes[1].set_xlabel("Число главных компонент")
axes[1].set_ylabel("Доля дисперсии")
axes[1].set_xticks(idx)
axes[1].legend(loc="upper left", framealpha=0.9)
plt.tight_layout()
plt.show()
```

```python
N_COMP = 0.85   # доля дисперсии, которую должны сохранить компоненты (середина диапазона 80–90%)

pca = PCA(n_components=N_COMP, random_state=RANDOM_STATE).fit(Z_train)
k = pca.n_components_
print(f"Выбрано главных компонент: {k} (сохранено {pca.explained_variance_ratio_.sum() * 100:.1f}% дисперсии)")

loadings = pd.DataFrame(pca.components_.T, index=X.columns, columns=[f"PC{i}" for i in range(1, k + 1)])
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.heatmap(loadings, annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax)
ax.set_title("Факторные нагрузки главных компонент")
plt.tight_layout()
plt.show()

# Проверка: компоненты не коррелированы, VIF = 1
pcs = pd.DataFrame(pca.transform(Z_train), columns=loadings.columns)
print("Максимальная |корреляция| между компонентами:",
      round(float(np.abs(pcs.corr().values - np.eye(k)).max()), 6))
print("VIF главных компонент:", compute_vif(pcs).round(3).tolist())
```

**Анализ результатов.**

* Собственные значения меняются плавно — от 1,21 до 0,76, **выраженного «локтя» на графике каменистой осыпи нет**. Первая компонента объясняет лишь 13,5 % дисперсии (для 9 полностью независимых признаков было бы 11,1 %).
* Общепринятый критерий Кайзера ($\lambda>1$; в материалах курса не приводится, использован как дополнительный ориентир) оставляет 4 компоненты (49,7 % дисперсии), что слишком мало. Порог накопленной дисперсии 85 % (середина диапазона 80–90 %, указанного в материалах курса) требует **8 компонент из 9** (91,5 %). Для дальнейших экспериментов выбрано **8 компонент**.
* Это прямое следствие отсутствия корреляций: каждая компонента почти совпадает с одним из исходных признаков или их небольшой смесью (в нагрузках нет ярко выраженной структуры; единственная заметная связь — `Solids` против `Sulfate` в PC1), поэтому сжать данные без потери информации нельзя.
* Проверка: компоненты не коррелированы между собой, VIF = 1.

### 4.6. Регрессионные модели на главных компонентах

Те же шесть моделей обучаются заново, но вместо исходных признаков на вход подаются 8 главных компонент (стандартизация → PCA → модель, всё внутри `Pipeline`).

```python
def pca_pipe(estimator):
    return Pipeline([("scaler", StandardScaler()),
                     ("pca", PCA(n_components=N_COMP, random_state=RANDOM_STATE)),
                     ("m", estimator)])

for name, (clf, grid) in logit_specs.items():
    results[("PCA", name)] = run_model(pca_pipe(clone(clf)), grid, scoring="neg_log_loss")
for name, (reg, grid, scoring) in linear_specs.items():
    results[("PCA", name)] = run_model(pca_pipe(clone(reg)), grid, scoring)

print("Логистические модели на главных компонентах:")
display(summary_table("PCA", logit_names).T)
print("Линейные модели (цель 0/1) на главных компонентах:")
display(summary_table("PCA", linear_names).T)
```

### 4.7. Сравнение моделей на исходных признаках и на главных компонентах

```python
rows = []
for fam, names in [("Логистическая", logit_names), ("Линейная (цель 0/1)", linear_names)]:
    for n in names:
        for space in ["Исходные", "PCA"]:
            r = results[(space, n)]
            rows.append({"Семейство": fam, "Модель": n, "Признаки": space,
                         **{f"CV {k}": r["cv"][k] for k in ["RMSE", "R2", "MAPE", "ROC-AUC"]},
                         **{f"Test {k}": r["test"][k] for k in ["RMSE", "R2", "MAPE", "ROC-AUC"]}})
cmp_df = pd.DataFrame(rows)
display(cmp_df.set_index(["Семейство", "Модель", "Признаки"]).round(4))

# Изменение метрик при переходе от исходных признаков к главным компонентам (PCA − исходные)
delta = (cmp_df[cmp_df["Признаки"] == "PCA"].set_index(["Семейство", "Модель"]).drop(columns="Признаки")
         - cmp_df[cmp_df["Признаки"] == "Исходные"].set_index(["Семейство", "Модель"]).drop(columns="Признаки"))
print("Изменение метрик (PCA − исходные признаки):")
display(delta.round(4))
```

```python
base = results[("Константа", "Baseline")]["test"]
metrics = [("RMSE", "RMSE (меньше — лучше)"), ("R2", "R² (больше — лучше)"),
           ("MAPE", "MAPE по классу 1 (меньше — лучше)"), ("ROC-AUC", "ROC-AUC (больше — лучше)")]

XLIMS = {"RMSE": (0, 0.6), "R2": (-0.05, 0.05), "MAPE": (0, 0.7), "ROC-AUC": (0.4, 0.7)}
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
for ax, (m, title) in zip(axes.ravel(), metrics):
    sns.barplot(data=cmp_df, y="Модель", x=f"Test {m}", hue="Признаки", ax=ax)
    ax.axvline(base[m], color="k", ls="--", lw=1, label="Baseline")
    ax.set_title(title)
    ax.set_ylabel("")
    ax.set_xlabel("")
    ax.set_xlim(*XLIMS[m])   # шкалы не обрезаны, чтобы не раздувать микроскопические различия
axes[0, 0].legend(fontsize=8)
for ax in axes.ravel()[1:]:
    if ax.get_legend() is not None:
        ax.get_legend().remove()
fig.suptitle("Метрики на тестовой выборке: исходные признаки и главные компоненты", fontsize=14)
plt.tight_layout()
plt.show()

plot_roc("PCA", logit_names, "ROC-кривые, логистические модели (главные компоненты)")
```

**Анализ результатов.**

* **Разницы практически нет.** При переходе к главным компонентам RMSE изменяется не более чем на 0,0008, $R^2$ — не более чем на 0,0034, CV ROC-AUC — на 0,0122 (и на ≤ 0,0011 у пяти моделей из шести). Все эти изменения лежат внутри разброса оценок между блоками кросс-валидации (± 0,02 по ROC-AUC).
* Наибольшее отличие — у логистической модели с L1-штрафом (ROC-AUC на тесте 0,505 → 0,472). Такое падение на 403 наблюдениях — шум: значение ниже 0,5 означает лишь, что модель на этой выборке не лучше случайного угадывания, а по CV разница всего 0,012.
* Объяснение теоретически простое. PCA — ортогональный поворот осей; модель без регуляризации на **всех** компонентах даёт те же предсказания, что и на исходных признаках, а штраф L2 (Ridge) поворотом не меняется. Единственное отличие создаёт отбрасывание девятой компоненты (8,5 % дисперсии), которая, как видно, не несёт полезного сигнала. L1-штраф зависит от выбора осей, но на данных без сигнала это тоже ни на что не влияет.
* Выигрыша PCA не даёт, потому что **мультиколлинеарности не было** (п. 4.1). Ни число обусловленности, ни устойчивость коэффициентов здесь не были проблемой.

Ниже показано, как меняется качество в зависимости от числа компонент.

```python
# Как меняется качество логистической регрессии (L2, C=1) в зависимости от числа компонент
ks = list(range(1, X.shape[1] + 1))
auc_by_k = []
for kk in ks:
    p = Pipeline([("scaler", StandardScaler()), ("pca", PCA(n_components=kk, random_state=RANDOM_STATE)),
                  ("m", make_logreg("l2"))])
    auc_by_k.append(cross_val_score(p, X_train, y_train, cv=outer_cv, scoring="roc_auc").mean())
auc_orig = cross_val_score(Pipeline([("scaler", StandardScaler()), ("m", make_logreg("l2"))]),
                           X_train, y_train, cv=outer_cv, scoring="roc_auc").mean()

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(ks, auc_by_k, "o-", label="PCA + логистическая регрессия")
ax.axhline(auc_orig, color="green", ls="--", label=f"Исходные признаки ({auc_orig:.3f})")
ax.axhline(0.5, color="k", ls=":", label="Случайное угадывание")
ax.set_xlabel("Число главных компонент")
ax.set_ylabel("ROC-AUC (кросс-валидация)")
ax.set_title("Качество модели в зависимости от числа компонент")
ax.set_xticks(ks)
ax.set_ylim(0.44, 0.6)
ax.legend()
plt.tight_layout()
plt.show()
```

При любом числе компонент $k<9$ ROC-AUC не превосходит значение для исходных признаков; при $k=9$ они совпадают (полный поворот осей), как и следует из теории. Подбор числа компонент ничего не улучшает.

### 4.8. Дополнительная проверка: заполнение пропусков вместо удаления

Удаление пропусков отняло 38,6 % данных. Проверим, не скрывает ли это сигнал: пропуски заполняются медианой (внутри `Pipeline`, без утечки), модель — логистическая с L2-штрафом.

```python
# Проверка: что изменится, если пропуски не удалять, а заполнить медианой (внутри pipeline — без утечки)
Xa, ya = df.drop(columns="Potability"), df["Potability"]
Xa_tr, Xa_te, ya_tr, ya_te = train_test_split(Xa, ya, test_size=0.2, random_state=RANDOM_STATE, stratify=ya)
data_all = (Xa_tr, ya_tr, Xa_te, ya_te)

pipe_imp = Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler()),
                     ("m", make_logreg("l2"))])
r_imp = run_model(pipe_imp, LOGIT_GRID, "neg_log_loss", data=data_all)
r_imp_base = run_model(Pipeline([("m", DummyClassifier(strategy="prior"))]), data=data_all)

cmp_imp = pd.DataFrame({
    f"Удаление пропусков (n={len(df_clean)}), L2":
        results[("Исходные", "Логистическая + L2 (Ridge)")]["test"],
    f"Заполнение медианой (n={len(df)}), L2": r_imp["test"],
    f"Заполнение медианой (n={len(df)}), Baseline": r_imp_base["test"],
}).T[["RMSE", "R2", "Accuracy", "F1", "ROC-AUC"]]
display(cmp_imp.round(4))
print("ROC-AUC при кросс-валидации (заполнение медианой, L2): "
      f"{r_imp['cv']['ROC-AUC']:.3f} ± {r_imp['cv_std']['ROC-AUC']:.3f}")
```

Точность модели (0,6098) по-прежнему **равна точности baseline**. ROC-AUC на тесте (0,552) выглядит чуть лучше, но кросс-валидация даёт **0,476 ± 0,019** — то есть ниже 0,5. Такое расхождение между двумя оценками — типичный признак отсутствия реального сигнала: случайные колебания по обе стороны от 0,5. Вывод не зависит от способа обработки пропусков.

### 4.9. Дополнительная проверка: нелинейная модель (вне рамок задания)

Чтобы понять, отсутствует ли зависимость вообще или линейные модели просто не подходят, обучен Random Forest на тех же выборках (без настройки гиперпараметров).

```python
# Вне рамок задания: проверка гипотезы о нелинейной зависимости (Random Forest, те же выборки и CV)
rf = Pipeline([("m", RandomForestClassifier(n_estimators=300, min_samples_leaf=3,
                                            random_state=RANDOM_STATE, n_jobs=-1))])
rf_cv = cross_val_score(rf, X_train, y_train, cv=outer_cv, scoring="roc_auc")
rf.fit(X_train, y_train)
rf_test = compute_scores(y_test, rf.predict_proba(X_test)[:, 1])

lr_cv = results[("Исходные", "Логистическая + L2 (Ridge)")]["cv"]["ROC-AUC"]
print(f"ROC-AUC (CV):   логистическая L2 = {lr_cv:.3f},  Random Forest = {rf_cv.mean():.3f} ± {rf_cv.std():.3f}")
print(f"Test: ROC-AUC = {rf_test['ROC-AUC']:.3f}, Accuracy = {rf_test['Accuracy']:.3f}, F1 = {rf_test['F1']:.3f}")

imp = pd.Series(rf.named_steps["m"].feature_importances_, index=X.columns).sort_values()
print("\nВажность признаков (Random Forest):")
print(imp.sort_values(ascending=False).round(3).to_string())
fig, ax = plt.subplots(figsize=(6.5, 3.8))
sns.barplot(x=imp.values, y=imp.index, ax=ax)
ax.set_title("Важность признаков (Random Forest)")
plt.tight_layout()
plt.show()
```

**Результат.** Random Forest достигает **CV ROC-AUC = 0,674 ± 0,019** (логистическая регрессия — 0,510) и точности 0,687 на тесте (baseline — 0,5955), F1 = 0,51. (Значение ROC-AUC на тесте, 0,728, выше кросс-валидационного; из-за малого размера теста надёжнее ориентироваться на CV.) Значит, **зависимость между признаками и питьевой пригодностью существует, но она нелинейная** (или определяется взаимодействием признаков), а важность признаков распределена довольно равномерно (от 0,084 до 0,160, наибольшая — у `Sulfate` и `ph`).

### 4.10. Анализ причин погрешности и способы повышения точности

**Почему линейные модели работают плохо:**
1. **Нет линейного сигнала.** Корреляции признаков с целью ≤ 0,034; модели не лучше константного прогноза. Главная причина — не предобработка и не выбор регуляризации, а **несоответствие класса модели структуре данных** (подтверждено Random Forest).
2. **Потеря данных.** Удаление пропусков сократило выборку на 38,6 %, что ухудшает оценку и может сместить распределение. Однако замена пропусков медианой картину не изменила.
3. **Малая тестовая выборка** (403 наблюдения): одиночная тестовая оценка ROC-AUC колеблется на несколько сотых, поэтому различия порядка 0,01–0,03 между моделями статистически не значимы.
4. **Несоответствие метрик задаче.** RMSE, $R^2$ и MAPE предназначены для регрессии. Для бинарной цели MAPE неопределён на классе 0, а $R^2$ и RMSE по вероятностям сообщают лишь то, что модель не лучше константы; содержательнее Accuracy, F1 и ROC-AUC.
5. **Происхождение метки.** В описании датасета не сказано, как именно определялась пригодность воды, поэтому неизвестно, насколько она вообще определяется измеренными показателями.

**Что можно сделать для повышения точности** (проверенные и непроверенные пункты отделены):

* *Проверено:* регуляризация (L1, L2, подбор $C$/$\alpha$ по сетке), стандартизация признаков, PCA с разным числом компонент, другой способ обработки пропусков — **существенного улучшения не дали**. Нелинейная модель (Random Forest) — дала (ROC-AUC 0,51 → 0,67).
* *Не проверялось, но логично попробовать:* настройка гиперпараметров ансамблей (градиентный бустинг, глубина и число деревьев); полиномиальные признаки и взаимодействия для линейных моделей; признаки на основе нормативов (из описания: pH 6,5–8,5, TDS ≤ 1000, хлорамины ≤ 4 мг/л, мутность ≤ 5 NTU и т. д.), например индикаторы «в пределах нормы»; более аккуратное заполнение пропусков (KNN, итеративное) вместо удаления; взвешивание классов (`class_weight`) и подбор порога классификации, если важны F1 или полнота по питьевому классу.

## 5. Заключение

1. Датасет Water Potability содержит 3276 наблюдений и 9 числовых признаков; целевая переменная бинарная (39 % питьевых), поэтому основной моделью выбрана логистическая регрессия (без штрафа, L1 и L2); дополнительно построены линейная регрессия, Lasso и Ridge на кодировке 0/1. После удаления пропусков (−38,6 %) осталось 2011 наблюдений.
2. **Мультиколлинеарность отсутствует:** максимальная попарная корреляция признаков 0,178, VIF от 1,002 до 1,050.
3. **Линейные модели не превосходят константный прогноз:** RMSE ≈ 0,491, $R^2\approx 0$, CV ROC-AUC ≈ 0,51, точность на тесте 0,5955 совпадает с baseline. Регуляризация не помогла (Lasso обнуляет почти все, а линейный Lasso — все коэффициенты).
4. **PCA не улучшил и не ухудшил качество:** по шкале каменистой осыпи выраженного «локтя» нет, для сохранения 85 % дисперсии требуется 8 компонент из 9; все метрики на главных компонентах отличаются от исходных в пределах случайного разброса (RMSE ±0,0008, $R^2$ ±0,0034). Это закономерно: PCA — поворот осей, а коррелированности признаков, которую нужно было устранять, в данных нет.
5. Главная причина низкого качества — **отсутствие линейной зависимости** цели от признаков. Дополнительная проверка показала, что зависимость существует, но нелинейная: Random Forest даёт CV ROC-AUC 0,67 против 0,51 у логистической регрессии. Поэтому для этой задачи следует переходить к нелинейным моделям, а не подбирать регуляризацию или число главных компонент.
6. Методический вывод: для бинарной цели метрики RMSE, $R^2$ и MAPE малоинформативны (MAPE неопределён на классе 0), их нужно дополнять Accuracy, F1 и ROC-AUC и обязательно сравнивать с константным baseline, а стандартизацию и PCA обучать только на обучающих данных, чтобы избежать утечки.
